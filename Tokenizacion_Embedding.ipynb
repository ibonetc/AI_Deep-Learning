## Entendiendo Tokenizacion y Embedding

```python
from transformers import AutoTokenizer, AutoModel
```

```python
texts=['Cancer diagnosis using Artificial Intelligence',
       'Artificial intelligence in oncology',
       'Machine learning applications in medicine']
```

```python
tokenizer= AutoTokenizer.from_pretrained('bert-base-uncased')
```

```python
tokens= tokenizer.tokenize(texts[1])
print(tokens)
```

```python
tokens= tokenizer(texts[1])
tokens
```

```python
for token in tokens['input_ids']:
  print(tokenizer.decode(token))
```

```python
tokenizer_spanish= AutoTokenizer.from_pretrained('dccuchile/bert-base-spanish-wwm-uncased')
```

```python
frase="Carolina es biomédica y está haciendo su tesis de doctorado en predicción de cáncer"
```

```python
tokens=tokenizer_spanish.tokenize(frase)
print(tokens)
```

```python
tokenizer_spanish.vocab_size
```

```python
tokens=tokenizer_spanish(frase)
tokens
```

```python
for t in tokens['input_ids']:
  print(tokenizer_spanish.decode([t]))
```

```python
tokenizer_spanish.vocab
```

## Embedding

```python
import torch
```

```python
model = AutoModel.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
```

```python
tokens=tokenizer_spanish.tokenize(frase)
print(tokens)
```

```python
inputs = tokenizer_spanish(frase, return_tensors="pt")
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
embeddings = outputs.last_hidden_state
```

jemplo: torch.Size([1, número_tokens, 768])

```python
print("Shape embeddings:", embeddings.shape)
```

```python
print(embeddings[0])
```

```python
embeddings_np=embeddings.numpy()
```
