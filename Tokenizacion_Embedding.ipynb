## Entendiendo Tokenizacion y Embedding

```python
from transformers import AutoTokenizer, AutoModel
```

```python
texts=['Cancer diagnosis using Artificial Intelligence',
       'Artificial intelligence in oncology',
       'Machine learning applications in medicine']
```

```python
tokenizer= AutoTokenizer.from_pretrained('bert-base-uncased')
```

```python
tokens= tokenizer.tokenize(texts[1])
print(tokens)
```

```python
tokens= tokenizer(texts[1])
tokens
```

```python
for token in tokens['input_ids']:
  print(tokenizer.decode(token))
```

## Embedding

```python
import torch
```

```python
model = AutoModel.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
```

```python
tokens=tokenizer_spanish.tokenize(frase)
print(tokens)
```

```python
inputs = tokenizer_spanish(frase, return_tensors="pt")
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
embeddings = outputs.last_hidden_state
```

```python
import torch
from transformers import BertModel, BertTokenizer
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Frase en español
sentence = "La inteligencia artificial es una tecnología que está transformando el mundo."
```

```python
# Carga del tokenizer y modelo multilingüe
tokenizer = BertTokenizer.from_pretrained('bert-base-multilingual-cased')
inputs = tokenizer(sentence, return_tensors="pt")
```

```python
# Tokenizar y extraer tokens
tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
```

```python
tokens
```

```python
# Carga del modelo
model = BertModel.from_pretrained('bert-base-multilingual-cased', output_attentions=True)
```

```python
# Obtener la salida del modelo (incluyendo las atenciones)
outputs = model(**inputs)
attentions = outputs.attentions
```

```python
# Seleccionamos la matriz de atención de la primera capa y la primera cabeza
attention_matrix = attentions[0][0][0].detach().numpy()  # Primera capa y primera cabeza
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(attention_matrix, annot=True, fmt=".2f", cmap="viridis", xticklabels=tokens, yticklabels=tokens)
plt.title("Matriz de Atención con Tokens Textuales")
plt.xlabel("Tokens")
plt.ylabel("Tokens")
plt.show()
```
